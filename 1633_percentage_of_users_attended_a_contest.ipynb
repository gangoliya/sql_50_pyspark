## Setup

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# Initialize Spark Session
spark = SparkSession.builder.appName("LeetCodeAggregation").getOrCreate()

# Users table data
users_data = [
    (6, "Alice"),
    (2, "Bob"),
    (7, "Alex")
]

# Register table data
register_data = [
    (215, 6), (215, 7), (215, 2),
    (208, 6), (208, 7),
    (210, 6), (210, 7), (210, 2),
    (207, 2), (207, 7),
    (209, 7)
]

# Create DataFrames
users_df = spark.createDataFrame(users_data, ["user_id", "user_name"])
register_df = spark.createDataFrame(register_data, ["contest_id", "user_id"])

users_df.show()
register_df.show()
```
