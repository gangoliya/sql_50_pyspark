## Setup
[Link to Question](https://leetcode.com/problems/number-of-unique-subjects-taught-by-each-teacher/description/?envType=study-plan-v2&envId=top-sql-50)

```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

# Initialize Spark Session
spark = SparkSession.builder.appName("LeetCode2356").getOrCreate()

# Input data from the LeetCode example
data = [
    (1, 2, 3),
    (1, 2, 4),
    (1, 3, 3),
    (2, 1, 1),
    (2, 2, 1),
    (2, 3, 1),
    (2, 4, 1)
]

# Define schema
columns = ["teacher_id", "subject_id", "dept_id"]

# Create DataFrame
teacher_df = spark.createDataFrame(data, schema=columns)
teacher_df.show()
```

## Code

```python
teacher_df.groupBy(F.col("teacher_id")).agg(F.countDistinct(F.col("subject_id")).alias("cnt")).show()
```
