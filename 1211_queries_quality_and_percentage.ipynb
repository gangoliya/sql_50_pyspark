## Setup

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType, IntegerType

# Initialize Spark Session
spark = SparkSession.builder.appName("LeetCode").getOrCreate()

# Define schema based on the problem description
schema = StructType([
    StructField("query_name", StringType(), True),
    StructField("result", StringType(), True),
    StructField("position", IntegerType(), True),
    StructField("rating", IntegerType(), True)
])

# Input data from the LeetCode example
data = [
    ("Dog", "Golden Retriever", 1, 5),
    ("Dog", "German Shepherd", 2, 5),
    ("Dog", "Mule", 200, 1),
    ("Cat", "Shirazi", 5, 2),
    ("Cat", "Siamese", 3, 3),
    ("Cat", "Sphynx", 7, 4)
]

# Create DataFrame
queries_df = spark.createDataFrame(data, schema)

queries_df.show()
```

## Code

```python
from pyspark.sql import functions as F

queries_df.groupBy("query_name").agg(F.round(F.avg(F.col("rating")/F.col("position")), 2).alias("quality"), \
                                    (F.round(F.sum(F.when(F.col("rating") < 3, 1).otherwise(0)) / F.count(F.col("rating")) * 100.0, 2)).alias("poor_query_percentage")).show()
```
